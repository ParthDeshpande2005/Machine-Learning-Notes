### find the last match played by virat kohli in Delhi

```python


ipl['all_players'] = ipl['Team1Players'] + ipl['Team2Players']
ipl.head(1) # we have added new coloumn named allplayers
```

```python
def did_kohli_play(players_list): #this will retur bool value according to if V kohli is in list.
    return 'V Kohli' in players_list
```

```python
# ex->
print(did_kohli_play(['R Sharma','S Dhawan']))
did_kohli_play(['V Kohli','R Sharma','S Dhawan'])
```

```python
ipl['did_kohli_play'] = ipl['all_players'].apply(did_kohli_play) # we addend new coloumn name did virat kohli play.

ipl.head(1)
```

```python
# now we have all the the match which virat played in delhi.
ipl[(ipl['City']=='Delhi') & (ipl['did_kohli_play']==True)].head(3)
```

```python
# finding the last match
ipl[(ipl['City']=='Delhi') & (ipl['did_kohli_play']==True)].drop_duplicates(subset=['City','did_kohli_play'],keep='first')
```

```python
points_df = pd.DataFrame(
    {
        '1st point':[(3,4),(-6,5),(0,0),(-10,1),(4,5)],
        '2nd point':[(-3,4),(0,0),(2,2),(10,10),(1,1)]
    }
)

points_df
```

```python
def euclidean(row):
    pt_A = row['1st point']
    pt_B = row['2nd point']

    return ((pt_A[0] - pt_B[0])**2 + (pt_A[1] - pt_B[1])**2)**0.5
```

```python
points_df['distance'] = points_df.apply(euclidean,axis=1) # default axis coloumn hota hai.

points_df
```
