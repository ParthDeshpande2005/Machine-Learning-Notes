```python
# []->list
# ()->tuple
# {}->dict,set
d={} #this is dict #if set is empty it is considered a dictionary.
s=set({}) #this is set
type(d) ,type(s)
```

```python
numbers=set([1,2,3,4,5,6])
print(numbers) #any order can be printed ,so we can't perform index operations.
#no duplicates in list.append

numbers.add(4)
numbers.add(13)
print(numbers)

numbers.remove(3)
print(numbers)
```

```python
#set operations->
A={1,2,3,4}
B={2,5,4,6}

union=A|B
print(union)

intersection=A&B
print(intersection)

difference=A-B
print(difference)

difference2=B-A
print(difference2)

sym_diff=A^B
print(sym_diff)
```

## Methods and Lambda Functions

### Points to Cover

- Defining functions uding def and understanding return values.
- Understanding parameters and arguments, including default arguments.
- Using lambda functions for small, anonymous functions.
- Higher-order functions like map(), filter(), sorted() with lambda functions.
- Scope and lifetime of variables in functions.

```python
def square_number(x):
    return x**2

square_number(22)
```

```python
def greet(name):
    return f"Hello {name}!!"

greet('bob')
```

```python
#method inside a method->
def outer_func():
    outer_var= 'outer func'

    def inner_func():
        inner_var='inner func'

        print(outer_var)
        print(inner_var)

    inner_func()
    print(outer_var)

outer_func()
```

```python
numbers2=list(range(10))
numbers2
```

```python
squares= list(map(lambda x: x**2,numbers2))
squares
```

```python
evens=list(filter(lambda x: x%2==0,squares))
evens
```

```python
fruits=['apple','mango','cherry','banana','orange']

print(sorted(fruits))

print(sorted(fruits,key=lambda x: x[-1])) #sort according to last index of the string.
```

## Exceptional Handling

### Points to Cover:

- Understanding exceptions and errors.
- Using try, except, else, and finally blocks.
- Raising exceptions using raise.
- Creating custom exceptions.
- Best practices for error handling.

```python
y=0.1 #0
x=1/y #could be anything

if y !=0:
    x=1/y

#square_number('12)

try:
    result=1/0
except Exception as e:
    print("You can't divide by zero")

    print(e)

# try, except, else, finally

try:
    number=int("3") #also try using 2 
except ValueError:
    print("that's not a valid number.")
    # raise(ValueError)  #this will raise the error.

else: #this only executes when except block does not run.meaning when there is no error.
    print(f"your number is {number}")

finally:
    print("I run allways")
```
