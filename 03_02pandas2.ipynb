## selecting rows from a DataFrame

- iloc - searches using index positions
- loc - searching using index lables

This is also valid in Series so going forward we use iloc and loc.

```python
stud ={
    'name':['nitish','ankit','rupesh','rishab','amit','ankita'],
    'iq':[100,90,120,80,60,0],
    'marks':[80,70,100,50,80,0],
    'package':[10,6,14,2,5,0]
}

newstudents = pd.DataFrame(stud)

newstudents.set_index('name',inplace=True)

newstudents
```

## filtering a DataFrame

```python
ipl.head(2)
```

```python
# find all the final winners

mask =  ipl['MatchNumber'] == 'Final'

new_df = ipl[mask]

print(new_df[['Season','WinningTeam']])

# or

ipl[ipl['MatchNumber']=='Final'][['Season','WinningTeam']]
```

```python
# how many super over finishes have occured

ipl[ipl['SuperOver']=='Y'].shape[0]
```

```python
# how many matches has csk won in kolkata

ipl[(ipl['WinningTeam']=='Chennai Super Kings') & (ipl['City']=='Kolkata')].shape[0]
```

```python
# toss winner is match winner in percentage
(ipl[ipl['TossWinner'] == ipl['WinningTeam']].shape[0] / ipl.shape[0])*100
```

```python
movies.head(2)
```

```python
# movies with rating higher than 8 and votes>10000

movies[(movies['imdb_rating']>8) & (movies['imdb_votes']>10000)].shape[0]
```

```python
# Action movies with rating higher than 7.5

# apply is used to use function.
# str is used to convert object to string.
# split is used to create the array from the string.
mask1 = movies['genres'].str.split('|').apply(lambda x:'Action' in x)

# or  movies['genres].str.contains('Action)

mask2 = movies['imdb_rating'] > 7.5

movies[mask1 & mask2].head(2)
```
