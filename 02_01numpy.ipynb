## What is numpy?

NumPy is the fundamental package for scientific computing in Python. It is a Python library that provides a multidimensional array object, various derived objects (such as masked arrays and matrices), and an assortment of routines for fast operations on arrays, including mathematical, logical, shape manipulation, sorting, selecting, I/O, discrete Fourier transforms, basic linear algebra, basic statistical operations, random simulation and much more.

At the core of the NumPy package, is the ndarray object. This encapsulates n-dimensional arrays of homogeneous data types


## Numpy Arrays Vs Python Sequences

- NumPy arrays have a fixed size at creation, unlike Python lists (which can grow dynamically). Changing the size of an ndarray will create a new array and delete the original.

- The elements in a NumPy array are all required to be of the same data type, and thus will be the same size in memory.

- NumPy arrays facilitate advanced mathematical and other types of operations on large numbers of data. Typically, such operations are executed more efficiently and with less code than is possible using Python’s built-in sequences.

- A growing plethora of scientific and mathematical Python-based packages are using NumPy arrays; though these typically support Python-sequence input, they convert such input to NumPy arrays prior to processing, and they often output NumPy arrays.

## Why we Need Numpy?

NumPy was created because Python's normal lists are not efficient or convenient for numerical computing, especially when working with large amounts of data. so Numpy uses the the C language to create the array in backend so it to be efficient and we also get pythons easy to use and learn syntax.

## Array Attributes


```python
a1=np.arange(10)
a2=np.arange(12,dtype=float).reshape(3,4)
a3=np.arange(8).reshape(2,2,2)
```

```python
# shape 
 
print(a1.shape)

print(a2.shape)

print(a3.shape)
```

```python
# size -> tell number of items in the array

print(a1.size)

print(a2.size)

print(a3.size)
```

```python
# itemsize -> tell the size of the stored atom

print(a1.itemsize) # int uses 8 byte

a4=np.arange(10,dtype=np.int32) # 32bit int uses 4byte

print(a4.itemsize)
```

```python
# dtype

print(a1.dtype)
print(a2.dtype)
print(a3.dtype)
print(a4.dtype)
```

## Array Functions

```python
a1 = np.random.random((3,3))
a1 = np.round(a1*100)
a1
```

```python
# max/min/sum/prod

print(np.max(a1))

print(np.min(a1))

print(np.sum(a1))

print(np.prod(a1))
```

```python
# axis
# 0 -> col and 1-> row

print(np.min(a1,axis=0)) # find min in every col
print(np.max(a1,axis=1)) # will give row wise maxi value.

```

```python
# mean/median/std/var

print(np.mean(a1,axis=1)) # find mean row wise

print(np.median(a1,axis=1)) # median 

print(np.std(a1,axis=1)) # standard deviation

print(np.var(a1,axis=1)) # variance
```

```python
# trigonometric fundtions

np.sin(a1)
```

```python
# dot product

a2 = np.arange(12).reshape(3,4)
a3 = np.arange(12,24).reshape(4,3) 

np.dot(a2,a3)
```

```python
# log and exponets

print(np.log(a1))

np.exp(a1)
```

```python
# round/floor/ceil

a5 = np.random.random((2,2))*100

print(np.round(a5))
print("")

print(np.floor(a5))
print("")

print(np.ceil(a5))
```

## Indexing and Slicing

```python
a1 = np.arange(10)
a2 = np.arange(12).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2)
```

```python
print(a1)

print(a1[0] , a1[-1]) # first and last.
```

```python
print(a2,"\n")

print(a2[1][2] , a2[-1][-1] , a2[0][0]) 
```

```python
print(a3,"\n")

print(a3[1][0][1], a3[-1][-1][-1], a3[0][0][0])
```

```python
# slicing

a1[2:7:2]
```

```python
a2
```

```python
print(a2[0,:],"\n") # printing 0th row

print(a2[:,2],"\n") # printing 2nd coloum

print(a2[1:3,1:3],"\n") # a:b mai last include nahi hota hai

print(a2[::2,::3],"\n") # print all four courner

# practice ->

print(a2[0:3:2,1:4:2],"\n") 

print(a2[1,::3],"\n") 

print(a2[:2,1:],"\n")
```

```python
a3 = np.arange(27).reshape(3,3,3) 

a3
```

```python
print(a3[1],"\n")

print(a3[::2],"\n")

print(a3[0,1,:],"\n")  # we cant use [][][] as this way we cant perform desired slicing this is good for knowing the item at the index

print(a3[1,:,1],"\n") 

print(a3[2,1:,1:],"\n")

print(a3[::2,0,::2],"\n")
```
