## np.where

The numpy.where() function returns the indices of elements in an input array where the given condition is satisfied.

```python
a
```

```python
# find all indicies with value greater than 50
np.where(a>50)
```

```python
# replace all values > 50 with 0

# np.where(condition,true,false)

np.where(a>50,0,a)
```

```python
# replace all the even with zero

np.where(a%2==0,0,a)
```

## Set functions

- np.union1d
- np.intersec1d
- np.setdiff1d
- np.setxor1d
- np.in1d

```python
m = np.array([1,2,3,4,5])
n = np.array([3,4,5,6,7])

np.union1d(m,n)
```

```python
np.intersect1d(m,n)
```

```python
print(np.setdiff1d(n,m)) #elements that are unique to n
np.setdiff1d(m,n) #elements that are unique to n
```

```python
np.setxor1d(n,m) #remove all common items only print unique elements
```

```python
m[np.in1d(m,1)]
```
